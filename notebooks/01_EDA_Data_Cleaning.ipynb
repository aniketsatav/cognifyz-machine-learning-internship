- New Delhi has the largest number of restaurant records.

## Cuisine
- The dataset contains 1,825 unique cuisine strings.
- Several records contain multiple cuisine categories in a single value.

## Cost
- Average Cost for two has a highly skewed distribution.
- Some unusually large values require investigation during data cleaning.

## Votes
- Votes are highly skewed, with a relatively small number of restaurants receiving very high vote counts.

# Day 3 — Data Cleaning Summary

## Issues Identified

1. 9 missing values were found in the Cuisines column.
2. No duplicate rows were found.
3. 2,148 restaurants had an Aggregate rating of 0 and were marked as Not rated.
4. 497 restaurants had both Latitude and Longitude equal to 0.
5. 18 restaurants had Average Cost for two equal to 0.
6. Several numerical variables showed highly skewed distributions.
7. Multiple cuisines were stored together in a single Cuisines field.
8. Rating color and Rating text are derived from Aggregate rating and should not be used as predictive features for Task 1.

## Cleaning Decisions

- Missing Cuisines values were replaced with "Unknown".
- Duplicate records were retained only after confirming that no duplicates existed.
- Restaurants with rating 0 were retained in the master dataset but excluded from the rated dataset used for rating prediction.
- Restaurants with invalid coordinates were retained in the master dataset but excluded from location-specific analysis.
- Zero values in Average Cost for two were treated as missing values for later preprocessing.
- Text columns were stripped of unnecessary leading/trailing whitespace.
- Task-specific datasets were created instead of applying the same transformations to all tasks.

## Next Step

The cleaned datasets will be used for feature engineering and machine learning tasks.
